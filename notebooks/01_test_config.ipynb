```python
# make the repo's src/ importable
import os, sys
repo_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.insert(0, repo_root)
print("repo root:", repo_root)
print("config exists:", os.path.exists(os.path.join(repo_root, "config", "config.yaml")))
```

```python
# Empty widget value = use the config default; a pipeline can override it.
dbutils.widgets.text("catalog", "", "Catalog override (optional)")
dbutils.widgets.text("load_date", "2026-11-26", "Load date (YYYY-MM-DD)")

catalog_override = dbutils.widgets.get("catalog").strip()
load_date = dbutils.widgets.get("load_date")

overrides = {"project": {"catalog": catalog_override}} if catalog_override else None
```

```python
# Cell 3: load config and print what it builds
from src import config_loader as cl

cfg = cl.load_config(os.path.join(repo_root, "config", "config.yaml"), overrides)

print("Pipeline order:", [r["name"] for r in cl.get_resources(cfg)])
for r in cl.get_resources(cfg):
    n = r["name"]
    print(f"\n{n}")
    print("  raw file :", cl.raw_file_path(cfg, n, load_date, 1))
    print("  bronze   :", cl.resource_table(cfg, "bronze", n))
    print("  silver   :", cl.resource_table(cfg, "silver", n))
print("\nrun log  :", cl.meta_table(cfg, "run_log"))
print("watermark:", cl.meta_table(cfg, "watermark"))
print("batch id :", cl.new_batch_id())
```
