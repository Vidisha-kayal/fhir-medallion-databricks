```python
import os, sys, json
from datetime import datetime, timezone
repo_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.insert(0, repo_root)

from src import config_loader as cl
from src import fhir_demo as fd

cfg = cl.load_config(os.path.join(repo_root, "config", "config.yaml"))
demo_dir = os.path.join(os.path.dirname(cl.raw_root(cfg)), "_demo")
state_file = os.path.join(demo_dir, "scd2_demo_patients.json")
os.makedirs(demo_dir, exist_ok=True)

if os.path.exists(state_file):
    raise Exception(f"Demo patients already created: {open(state_file).read()}")

run_tag = f"scd2-demo-{cl.utc_now():%Y%m%d}"
alpha = fd.create_test_patient(cfg, "ScdDemoAlpha", "Test", "Pune", run_tag)
bravo = fd.create_test_patient(cfg, "ScdDemoBravo", "Test", "Delhi", run_tag)

load_date = datetime.fromisoformat(alpha["last_updated"]).astimezone(timezone.utc).strftime("%Y-%m-%d")
with open(state_file, "w") as f:
    json.dump({"run_tag": run_tag, "alpha": alpha, "bravo": bravo, "load_date": load_date}, f)

print("alpha:", alpha)
print("bravo:", bravo)
print(">>> USE THIS LOAD DATE (UTC) IN ALL STEPS BELOW:", load_date)
```
